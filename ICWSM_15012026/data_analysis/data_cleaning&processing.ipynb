```python
import pandas as pd
import numpy as np
import os
import pyarrow.parquet as pq
```

```python
def read_parquet_with_checks(file_path):
    print(f"File size: {os.path.getsize(file_path)} bytes")

    # Try to read just metadata
    try:
        metadata = pq.read_metadata(file_path)
        print("Metadata read successfully")
        print(f"Number of rows: {metadata.num_rows}")
        print(f"Number of columns: {metadata.num_columns}")
    except Exception as e:
        print(f"Error reading metadata: {e}")

    try:
      df= pd.read_parquet(
        file_path,
        engine='fastparquet'  # Instead of pyarrow
        )
    except Exception as e:
      print(f"Fastparquet error: {e}")    
    return df  
```

```python
import geopandas as gpd
gdf = gpd.read_file("/Users/i2scg/Documents/PHD/sm_project/btw25_geometrie_wahlkreise_shp_geo/btw25_geometrie_wahlkreise_shp_geo.shp")

#rename columns
gdf.rename(columns={'WKR_NR':'constituency_nr','WKR_NAME':'constituency' , 'LAND_NAME':'state', 'LAND_NR':'state_nr'}, inplace=True)
gdf.info()
```

```python

# Create points GeoDataFrame with explicit CRS
merged_gdf = gpd.GeoDataFrame(
    merged_df, 
    geometry=gpd.points_from_xy(merged_df.longitude, merged_df.latitude), 
    crs="EPSG:4326"  # WGS84 (lat/lon)
)

print("Points CRS:", merged_gdf.crs)
print("Points bounds:", merged_gdf.total_bounds)

# Ensure both have the same CRS
if gdf.crs != merged_gdf.crs:
    print(f"CRS mismatch! Converting points to shapefile CRS: {gdf.crs}")
    merged_gdf = merged_gdf.to_crs(gdf.crs)
```

```python
# First try intersects instead of within
merged_gdf = gpd.sjoin(merged_gdf, gdf, how='left', predicate='intersects')

# If there are still issues, use this comprehensive approach:
if merged_gdf['index_right'].isna().sum() > 0:
    # For remaining points, use nearest neighbor
    from scipy.spatial import cKDTree
    
    # Get centroids and build KDTree
    centroids = np.array([[p.x, p.y] for p in gdf.geometry.centroid])
    tree = cKDTree(centroids)
    
    non_matching_idx = merged_gdf['index_right'].isna()
    non_matching_points = np.array([[p.x, p.y] for p in merged_gdf.loc[non_matching_idx, 'geometry']])
    
    if len(non_matching_points) > 0:
        distances, indices = tree.query(non_matching_points)
        
        for i, idx in enumerate(merged_gdf[non_matching_idx].index):
            nearest = gdf.iloc[indices[i]]
            merged_gdf.loc[idx, ['constituency_nr', 'constituency', 'state', 'state_nr']] = \
                nearest[['constituency_nr', 'constituency', 'state', 'state_nr']].values
```

```python
pools_qgis=pools_df[['place_id', 'place_name', 'pool_type', 'latitude_gmaps', 'longitude_gmaps', 
          'total_reviews', 'place_rating', 'city_admin']]
```

```python
pools_gdf = gpd.GeoDataFrame(
    pools_qgis, 
    geometry=gpd.points_from_xy(pools_qgis.longitude_gmaps, pools_qgis.latitude_gmaps),
    crs="EPSG:4326"
)
# Ensure both GeoDataFrames use the same CRS before spatial join
if shapefile_con.crs != pools_gdf.crs:
    shapefile_con_reproj = shapefile_con.to_crs(pools_gdf.crs)
else:
    shapefile_con_reproj = shapefile_con


# First try within join
pools_within = gpd.sjoin(pools_gdf, shapefile_con_reproj, how="inner", predicate='within')

# For points that didn't match, try intersects
remaining_points = pools_gdf[~pools_gdf.index.isin(pools_within.index)]
pools_intersects = gpd.sjoin(remaining_points, shapefile_con_reproj, how="inner", predicate='intersects')

# Combine results
pools_qgis_gdf = pd.concat([pools_within, pools_intersects], ignore_index=True)

# If there are still unmatched points, do a nearest join as last resort
if len(pools_qgis_gdf) < len(pools_gdf):
    remaining_points = pools_gdf[~pools_gdf.index.isin(pools_qgis_gdf.index)]
    pools_nearest = gpd.sjoin_nearest(remaining_points, shapefile_con_reproj, how="left", max_distance=1000)
    pools_qgis_gdf = pd.concat([pools_qgis_gdf, pools_nearest], ignore_index=True)

pools_qgis_gdf.info()    
```

```python
final_pools_qgis = pd.merge( pools_qgis_gdf, demographics_df, on=['constituency_number',
                            'constituency_name','land_name' ], how='right')
final_pools_qgis.drop(columns=['index_right'], inplace=True)
final_pools_qgis.info()
```
