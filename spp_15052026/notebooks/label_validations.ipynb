```python
import pandas as pd
from sklearn.metrics import classification_report

# ── Typo + normalization corrections ─────────────────────────────────────────

context_corrections = {
    "inident":     "incident",
    "generalsed":  "generalised",
    "generalized": "generalised",
}

connotation_corrections = {
    "negativve":        "negative",
    "negatvie":         "negative",
    "posiitve":         "positive",
    "positve":          "positive",
    # normalize label order
    "positive, negative": "negative, positive",
    "neutral, negative":  "negative, neutral",
    "positive, neutral":  "neutral, positive",
}

# ── Cleaning function ─────────────────────────────────────────────────────────

def clean_label(series, corrections=None):
    s = (series
        .astype(str)
        .str.strip()
        .str.lower()
        .str.replace(r'\s*,\s*', ', ', regex=True)  # normalize comma spacing
    )
    if corrections:
        s = s.replace(corrections)
    return s

# ── Evaluation function ───────────────────────────────────────────────────────

def evaluate(df, model_col, human_col, task_name, corrections=None):
    pair = df[[model_col, human_col]].dropna()

    y_true = clean_label(pair[human_col], corrections)
    y_pred = clean_label(pair[model_col], corrections)

    print(f"\n{'='*55}")
    print(f"Task: {task_name}  ({len(y_true)} samples)")
    print(f"{'='*55}")
    print(classification_report(y_true, y_pred, zero_division=0))

# ── Run ───────────────────────────────────────────────────────────────────────

evaluate(context,     "context",     "context_human",     "Context",     context_corrections)
evaluate(connotation, "connotation", "connotation_human", "Connotation", connotation_corrections)
```
